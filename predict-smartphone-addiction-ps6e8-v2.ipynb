```python
def build_features(df, cat_cols, num_cols, category_map=None, fit=False, median_dict=None):
    if category_map is None:
        category_map = {}
    if median_dict is None:
        median_dict = {}
    df = df.copy()
    new_cat = []
    new_num = []

    # Missing indicators 
    for col in num_cols:
        miss_col = f"miss_{col}"
        df[miss_col] = df[col].isnull().astype(np.int32)
        new_cat.append(miss_col)

    # Row-wise missing count
    df["missing_count"] = df[num_cols].isnull().sum(axis=1).astype(np.float32)
    new_num.append("missing_count")

    # Fill missing values
    for col in cat_cols:
        df[col] = df[col].fillna("missing")
    for col in num_cols:
        fill_val = median_dict.get(col, df[col].median())
        df[col] = df[col].fillna(fill_val)

    # Total screen time & weekend vs daily difference
    if "daily_screen_time_hours" in df.columns and "weekend_screen_time" in df.columns:
        df["total_screen"] = df["daily_screen_time_hours"] + df["weekend_screen_time"]
        df["weekend_diff"] = df["weekend_screen_time"] - df["daily_screen_time_hours"]
        new_num += ["total_screen", "weekend_diff"]
    if "social_media_hours" in df.columns and "daily_screen_time_hours" in df.columns:
        df["social_ratio"] = df["social_media_hours"] / (df["daily_screen_time_hours"] + 1e-6)
        new_num.append("social_ratio")
    if "gaming_hours" in df.columns and "daily_screen_time_hours" in df.columns:
        df["gaming_ratio"] = df["gaming_hours"] / (df["daily_screen_time_hours"] + 1e-6)
        new_num.append("gaming_ratio")
    activity = [c for c in ["social_media_hours", "gaming_hours", "work_study_hours"] if c in df.columns]
    if len(activity) == 3:
        df["activity_total"] = df[activity].sum(axis=1)
        df["activity_ratio"] = df["activity_total"] / (df["daily_screen_time_hours"] + 1e-6)
        new_num += ["activity_total", "activity_ratio"]
    if "daily_screen_time_hours" in df.columns and "sleep_hours" in df.columns:
        df["screen_sleep_ratio"] = df["daily_screen_time_hours"] / (df["sleep_hours"] + 1e-6)
        df["sleep_deficit"] = (8.5 - df["sleep_hours"]).clip(lower=0)
        new_num += ["screen_sleep_ratio", "sleep_deficit"]
    if "notifications_per_day" in df.columns and "app_opens_per_day" in df.columns:
        df["engagement"] = df["notifications_per_day"] + df["app_opens_per_day"]
        df["notif_per_open"] = df["notifications_per_day"] / (df["app_opens_per_day"] + 1e-6)
        new_num += ["engagement", "notif_per_open"]
    if "notifications_per_day" in df.columns:
        df["log_notifications"] = np.log1p(df["notifications_per_day"])
        new_num.append("log_notifications")
    if "app_opens_per_day" in df.columns:
        df["log_app_opens"] = np.log1p(df["app_opens_per_day"])
        new_num.append("log_app_opens")
    if "daily_screen_time_hours" in df.columns and "app_opens_per_day" in df.columns:
        df["screen_per_open"] = (df["daily_screen_time_hours"] * 60) / (df["app_opens_per_day"] + 1e-6)
        new_num.append("screen_per_open")
    # Stress-level numeric encoding
    if "stress_level" in df.columns:
        stress_map = {"Low": 0, "Medium": 1, "High": 2, "missing": 1}
        stress = df["stress_level"].map(stress_map).fillna(1).astype(np.float32)
        if "daily_screen_time_hours" in df.columns:
            df["screen_x_stress"] = df["daily_screen_time_hours"] * stress
            new_num.append("screen_x_stress")
        if "social_media_hours" in df.columns:
            df["social_x_stress"] = df["social_media_hours"] * stress
            new_num.append("social_x_stress")
        if "sleep_hours" in df.columns:
            df["sleep_x_stress"] = df["sleep_hours"] * stress
            new_num.append("sleep_x_stress")

    # Factorize original categorical columns
    for col in cat_cols:
        if fit:
            codes, uniques = pd.factorize(df[col], sort=True)
            category_map[col] = uniques
        else:
            uniques = category_map[col]
            code_map = {v: i for i, v in enumerate(uniques)}
            codes = df[col].map(code_map).fillna(-1).astype(np.int32)
        df[col] = codes.astype(np.int32)

    # Convert numerical columns into categorical form
    for col in num_cols:
        cat_name = f"{col}_cat"
        if fit:
            codes, uniques = pd.factorize(df[col], sort=True)
            category_map[cat_name] = uniques
        else:
            uniques = category_map[cat_name]
            code_map = {v: i for i, v in enumerate(uniques)}
            codes = df[col].map(code_map).fillna(-1).astype(np.int32)
        df[cat_name] = codes.astype(np.int32)
        new_cat.append(cat_name)

    # Quantile binning of key numerical features
    bin_targets = {
        "daily_screen_time_hours": 10,
        "social_media_hours": 10,
        "sleep_hours": 8,
    }
    for col, n_bins in bin_targets.items():
        if col not in df.columns:
            continue
        bin_name = f"{col}_qbin"
        if fit:
            kb = KBinsDiscretizer(n_bins=n_bins, encode="ordinal", strategy="quantile", subsample=None)
            binned = kb.fit_transform(df[[col]]).ravel().astype(np.int32)
            category_map[bin_name] = kb
        else:
            kb = category_map[bin_name]
            binned = kb.transform(df[[col]]).ravel().astype(np.int32)
        df[bin_name] = binned
        new_cat.append(bin_name)

    # Frequency encoding for original categorical columns
    for col in cat_cols:
        freq_name = f"{col}_freq"
        if fit:
            freq = df[col].value_counts(normalize=True)
            category_map[freq_name] = freq
            df[freq_name] = df[col].map(freq).astype(np.float32)
        else:
            freq = category_map[freq_name]
            df[freq_name] = df[col].map(freq).fillna(0.0).astype(np.float32)
        new_num.append(freq_name)
    return df, new_cat, new_num
```

# <p style="background:linear-gradient(120deg,#453548 0%,#60495D 24%,#806477 48%,#70837C 73%,#9A8B6B 100%);color:#FFFDF8;font-family:'Trebuchet MS','Verdana',sans-serif;font-size:24px;text-align:center;padding:22px 20px;border-radius:18px;width:78%;margin:34px auto;font-weight:800;letter-spacing:1.2px;line-height:1.3;text-shadow:0 2px 4px rgba(0,0,0,0.3);border:3px solid #E7D9C9;box-shadow:0 0 0 3px #493B4A,0 0 0 7px #C5A66A,0 10px 22px rgba(55,43,55,0.28),inset 0 2px 0 rgba(255,255,255,0.2);outline:3px solid #8E6B86;outline-offset:9px;">Let’s Build the Neural Tabular Classification Model</p>

```python
# Neural Tabular Classification Model
# NumericalPreprocessor
class NumericalPreprocessor(BaseEstimator, TransformerMixin):
    def __init__(self, transforms):
        self.transforms = [t for t in transforms if t in ("median_center", "robust_scale", "smooth_clip", "l2_normalize")]

    def fit(self, X, y=None):
        if "median_center" in self.transforms or "robust_scale" in self.transforms:
            self.median_ = np.median(X, axis=0)
            q75 = np.quantile(X, 0.75, axis=0)
            q25 = np.quantile(X, 0.25, axis=0)
            iqr = q75 - q25
            zero_mask = iqr == 0.0
            iqr[zero_mask] = 0.5 * (X.max(axis=0)[zero_mask] - X.min(axis=0)[zero_mask])
            self.iqr_scale_ = 1.0 / (iqr + 1e-30)
            self.iqr_scale_[iqr == 0.0] = 0.0
        return self

    def transform(self, X):
        X = X.astype(np.float32).copy()
        for t in self.transforms:
            if t == "median_center":
                X -= self.median_[None, :]
            elif t == "robust_scale":
                X *= self.iqr_scale_[None, :]
            elif t == "smooth_clip":
                X = X / np.sqrt(1.0 + (X / 3.0) ** 2)
            elif t == "l2_normalize":
                norms = np.linalg.norm(X, axis=1, keepdims=True)
                X /= np.where(norms == 0, 1.0, norms)
        return X

# CategoricalFeatureLayer
class CategoricalFeatureLayer(nn.Module):
    def __init__(self, n_ens, cat_dims, embed_dim=8, onehot_thresh=8):
        super().__init__()
        self.n_ens = n_ens
        self.cat_dims = cat_dims
        self.onehot_idx = []
        self.embed_layers = nn.ModuleList()
        self.embed_idx = []

        for i, dim in enumerate(cat_dims):
            if dim <= onehot_thresh:
                self.onehot_idx.append(i)
            else:
                emb = nn.ModuleList([nn.Embedding(dim, embed_dim) for _ in range(n_ens)])
                self.embed_layers.append(emb)
                self.embed_idx.append(i)

    def forward(self, x):
        B, E, _ = x.shape
        outs = []

        if self.onehot_idx:
            oh_x = x[:, :, self.onehot_idx]
            dims = [self.cat_dims[i] for i in self.onehot_idx]
            total = sum(dims)
            encoded = torch.zeros(B, E, total, device=x.device)
            start = 0
            for j, d in enumerate(dims):
                pos = oh_x[:, :, j:j+1].long()
                encoded.scatter_(2, pos + start, 1.0)
                start += d
            outs.append(encoded)

        for emb_list, feat_i in zip(self.embed_layers, self.embed_idx):
            emb_parts = []
            for m in range(self.n_ens):
                idx = x[:, m, feat_i:feat_i+1].long()
                emb_parts.append(emb_list[m](idx))
            outs.append(torch.cat(emb_parts, dim=1))

        return torch.cat(outs, dim=2) if outs else torch.zeros(B, E, 0, device=x.device)

# ScalingLayer
class ScalingLayer(nn.Module):
    def __init__(self, n_ens, n_features):
        super().__init__()
        self.scale = nn.Parameter(torch.ones(n_ens, n_features))

    def forward(self, x):
        return x * self.scale[None, :, :]

# NTPLinear
class NTPLinear(nn.Module):
    def __init__(self, n_ens, in_features, out_features, bias=True):
        super().__init__()
        self.in_features = in_features
        self.weight = nn.Parameter(torch.randn(n_ens, in_features, out_features))
        self.bias = nn.Parameter(torch.randn(n_ens, out_features)) if bias else None

    def forward(self, x):
        x = torch.einsum("bki,kio->bko", x, self.weight) / math.sqrt(self.in_features)
        if self.bias is not None:
            x = x + self.bias
        return x

# PBLDEmbedding
class PBLDEmbedding(nn.Module):
    def __init__(self, n_ens, n_features, hidden_dim=16, out_dim=4,
                 freq_scale=0.1, activation=nn.GELU):
        super().__init__()
        self.n_ens = n_ens
        self.n_features = n_features
        self.out_dim = out_dim

        self.w1 = nn.Parameter(torch.randn(n_ens, n_features, hidden_dim) * freq_scale)
        self.b1 = nn.Parameter(torch.randn(n_ens, n_features, hidden_dim))
        self.w2 = nn.Parameter(torch.randn(n_ens, n_features, hidden_dim, out_dim - 1) / math.sqrt(hidden_dim))
        self.b2 = nn.Parameter(torch.zeros(n_ens, n_features, out_dim - 1))
        self.act = activation()
        nn.init.uniform_(self.b1, -math.pi, math.pi)

    def forward(self, x):
        # x: (B, E, F)
        periodic = torch.cos(2 * math.pi * (x.unsqueeze(-1) * self.w1.unsqueeze(0) + self.b1.unsqueeze(0)))
        transformed = self.act(
            torch.einsum("bkfh,kfhd->bkfd", periodic, self.w2) + self.b2.unsqueeze(0)
        )
        feat = torch.cat([x.unsqueeze(-1), transformed], dim=-1)
        return feat.flatten(start_dim=2)

# RealMLP
class RealMLP(nn.Module):
    def __init__(self, output_dim, cat_dims, n_numerical, cfg):
        super().__init__()
        n_ens = cfg["n_ens"]
        self.n_ens = n_ens

        self.cate = CategoricalFeatureLayer(
            n_ens=n_ens,
            cat_dims=cat_dims,
            embed_dim=cfg["embed_dim"],
            onehot_thresh=cfg["onehot_thresh"],
        )
        self.num_embed = PBLDEmbedding(
            n_ens=n_ens,
            n_features=n_numerical,
            hidden_dim=cfg["pbld_hidden_dim"],
            out_dim=cfg["pbld_out_dim"],
            freq_scale=cfg["pbld_freq_scale"],
            activation=cfg["pbld_activation"],
        )

        num_dim = n_numerical * cfg["pbld_out_dim"]
        cat_dim = sum(d if d <= cfg["onehot_thresh"] else cfg["embed_dim"] for d in cat_dims)
        total_dim = num_dim + cat_dim

        layers = []
        if cfg["add_front_scale"]:
            layers.append(ScalingLayer(n_ens, total_dim))

        self.dropout_modules = []
        in_dim = total_dim
        for i, h in enumerate(cfg["hidden_dims"]):
            linear = NTPLinear(n_ens, in_dim, h)
            if i == 0:
                self.first_linear = linear
            drop = nn.Dropout(cfg["dropout"])
            self.dropout_modules.append(drop)
            layers += [linear, cfg["activation"](), drop]
            in_dim = h

        self.hidden = nn.Sequential(*layers)
        self.head = NTPLinear(n_ens, in_dim, output_dim)

    def forward(self, x_num, x_cat):
        x_num = x_num.unsqueeze(1).expand(-1, self.n_ens, -1)
        x_cat = x_cat.unsqueeze(1).expand(-1, self.n_ens, -1)
        x_num = self.num_embed(x_num)
        x_cat = self.cate(x_cat)
        x = torch.cat([x_num, x_cat], dim=2)
        x = self.hidden(x)
        x = self.head(x)
        return F.softmax(x, dim=2)
```

# <p style="background:linear-gradient(120deg,#453548 0%,#60495D 24%,#806477 48%,#70837C 73%,#9A8B6B 100%);color:#FFFDF8;font-family:'Trebuchet MS','Verdana',sans-serif;font-size:24px;text-align:center;padding:22px 20px;border-radius:18px;width:78%;margin:34px auto;font-weight:800;letter-spacing:1.2px;line-height:1.3;text-shadow:0 2px 4px rgba(0,0,0,0.3);border:3px solid #E7D9C9;box-shadow:0 0 0 3px #493B4A,0 0 0 7px #C5A66A,0 10px 22px rgba(55,43,55,0.28),inset 0 2px 0 rgba(255,255,255,0.2);outline:3px solid #8E6B86;outline-offset:9px;">Model Training, Validation & Prediction</p>

```python
# Applies different scheduling strategies to adjust a parameter during training.
def apply_schedule(init_value, progress, sched, flat_ratio=0.3):
    if sched == "constant":
        return init_value
    if sched == "cos":
        return init_value * (math.cos(math.pi * progress) + 1) / 2
    if sched == "flat_cos":
        if progress < flat_ratio:
            return init_value
        t = (progress - flat_ratio) / (1 - flat_ratio)
        return init_value * (math.cos(math.pi * t) + 1) / 2
    if sched == "expm4t":
        return init_value * math.exp(-4 * progress)
    raise ValueError(f"Unknown schedule: {sched}")

# Groups model parameter
def get_param_groups(model, p):
    first_w_id = id(model.first_linear.weight)
    scale_p, pbld_p, first_w, other_w, bias_p = [], [], [], [], []
    for name, param in model.named_parameters():
        if "num_embed" in name:
            pbld_p.append(param)
        elif "scale" in name:
            scale_p.append(param)
        elif id(param) == first_w_id:
            first_w.append(param)
        elif "bias" in name:
            bias_p.append(param)
        else:
            other_w.append(param)
    LR, WD = p["lr"], p["weight_decay"]
    return [
        {"params": scale_p,  "lr": LR * p["lr_scale_mult"],        "weight_decay": WD * p["wd_scale_mult"]},
        {"params": pbld_p,   "lr": LR * p["pbld_lr_factor"],       "weight_decay": WD},
        {"params": first_w,  "lr": LR * p["first_layer_lr_factor"],"weight_decay": WD * p["first_layer_wd_factor"]},
        {"params": other_w,  "lr": LR,                             "weight_decay": WD},
        {"params": bias_p,   "lr": LR * p["lr_bias_mult"],         "weight_decay": WD * p["wd_bias_mult"]},
    ]

# Computes cross-entropy loss
def smooth_ce(y_true, y_pred, ls=0.0, class_weights=None):
    n_classes = y_pred.size(1)
    y_smooth = torch.full_like(y_pred, ls / n_classes)
    y_smooth.scatter_(1, y_true.unsqueeze(1), 1.0 - ls + ls / n_classes)
    loss = -(y_smooth * torch.log(y_pred.clamp(1e-15, 1))).sum(dim=1)
    if class_weights is not None:
        w = class_weights[y_true]
        return (loss * w).sum() / w.sum()
    return loss.mean()

# RealMLP Classification Model
class RealMLPClassifier(BaseEstimator):
    def __init__(self, **kwargs):
        self.params = {**CFG, **kwargs}
    # Train the RealMLP model
    def fit(self, X_train, y_train, X_val, y_val, cat_col_names=None, X_test=None):
        p = self.params
        dev = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        verbose = p["verbosity"]
        cat_col_names = cat_col_names or []
        num_col_names = [c for c in X_train.columns if c not in cat_col_names]
        Xtr_num = X_train[num_col_names].values.astype(np.float32)
        Xva_num = X_val[num_col_names].values.astype(np.float32)
        Xtr_cat = X_train[cat_col_names].values.astype(np.int64)
        Xva_cat = X_val[cat_col_names].values.astype(np.int64)
        ytr = np.asarray(y_train)
        yva = np.asarray(y_val)

        # Numerical preprocessing
        self.preproc_ = NumericalPreprocessor(p["tfms"])
        self.preproc_.fit(Xtr_num)
        Xtr_num = self.preproc_.transform(Xtr_num)
        Xva_num = self.preproc_.transform(Xva_num)
        self.cat_cols_ = cat_col_names
        self.num_cols_ = num_col_names

        # Categorical dimensions 
        if cat_col_names:
            all_cat = [Xtr_cat, Xva_cat]
            if X_test is not None:
                all_cat.append(X_test[cat_col_names].values.astype(np.int64))
            cat_dims = (np.concatenate(all_cat, axis=0).max(axis=0) + 1).tolist()
        else:
            cat_dims = []
        self.cat_dims_ = cat_dims
        if cat_dims:
            cat_max = np.array(cat_dims) - 1
            Xtr_cat = np.clip(Xtr_cat, 0, cat_max)
            Xva_cat = np.clip(Xva_cat, 0, cat_max)

        # Class weights
        classes = np.unique(ytr)
        self.classes_ = classes
        cw = compute_class_weight("balanced", classes=classes, y=ytr)
        class_weights = torch.as_tensor(cw, dtype=torch.float32, device=dev)

        # Model
        n_classes = len(classes)
        self.model_ = RealMLP(
            output_dim=n_classes,
            cat_dims=cat_dims,
            n_numerical=Xtr_num.shape[1],
            cfg=p,
        ).to(dev)
        param_groups = get_param_groups(self.model_, p)
        for g in param_groups:
            g["lr_base"] = g["lr"]
        optimizer = torch.optim.AdamW(param_groups, betas=(p["mom"], p["sq_mom"]))
        Xtn = torch.as_tensor(Xtr_num, dtype=torch.float32, device=dev)
        Xtc = torch.as_tensor(Xtr_cat, dtype=torch.long, device=dev)
        ytt = torch.as_tensor(ytr, dtype=torch.long, device=dev)
        Xvn = torch.as_tensor(Xva_num, dtype=torch.float32, device=dev)
        Xvc = torch.as_tensor(Xva_cat, dtype=torch.long, device=dev)
        n_ens = p["n_ens"]
        train_bs = p["train_bs"]
        eval_bs = p["eval_bs"]
        epochs = p["epochs"]
        total_steps = epochs * len(ytr)
        order = np.arange(len(ytr))
        best_score = -np.inf
        best_epoch = 0
        best_val_probs = None
        best_state = None
        ema_state = None
        if p["ema_decay"] > 0:
            ema_state = {k: v.detach().clone() for k, v in self.model_.state_dict().items()}

        for epoch in range(epochs):
            self.model_.train()
            for start in range(0, len(ytr), train_bs):
                progress = (epoch * len(ytr) + start) / total_steps
                idx = order[start:start + train_bs]
                for g in optimizer.param_groups:
                    g["lr"] = apply_schedule(g["lr_base"], progress, p["lr_sched"], p["flat_ratio"])
                optimizer.zero_grad()
                pred = self.model_(Xtn[idx], Xtc[idx])
                ls = apply_schedule(p["ls_eps"], progress, p["ls_eps_sched"], p["flat_ratio"])
                drop = apply_schedule(p["dropout"], progress, p["p_drop_sched"], p["flat_ratio"])
                for dm in self.model_.dropout_modules:
                    dm.p = drop
                loss = smooth_ce(
                    ytt[idx].repeat_interleave(n_ens),
                    pred.reshape(-1, n_classes),
                    ls=ls,
                    class_weights=class_weights,
                )
                loss.backward()
                torch.nn.utils.clip_grad_norm_(self.model_.parameters(), p["grad_clip"])
                optimizer.step()
                if ema_state is not None:
                    with torch.no_grad():
                        for k, v in self.model_.state_dict().items():
                            if torch.is_floating_point(v):
                                ema_state[k].mul_(p["ema_decay"]).add_(v.detach(), alpha=1 - p["ema_decay"])
                            else:
                                ema_state[k].copy_(v)

            np.random.shuffle(order)

            # Validation with EMA
            self.model_.eval()
            if ema_state is not None:
                live = {k: v.detach().clone() for k, v in self.model_.state_dict().items()}
                self.model_.load_state_dict(ema_state, strict=True)
            with torch.no_grad():
                val_probs = np.concatenate([
                    self.model_(Xvn[s:s+eval_bs], Xvc[s:s+eval_bs]).mean(dim=1).cpu().numpy()
                    for s in range(0, len(yva), eval_bs)
                ], axis=0)
            score = roc_auc_score(yva, val_probs[:, 1]) if n_classes == 2 else \
                    balanced_accuracy_score(yva, np.argmax(val_probs, axis=1))
            improved = score > best_score
            if improved:
                best_score = score
                best_epoch = epoch + 1
                best_val_probs = val_probs.copy()
                src = ema_state if ema_state is not None else self.model_.state_dict()
                best_state = {k: v.detach().clone() for k, v in src.items()}
            if verbose >= 2:
                mark = " *" if improved else ""
                print(f"  epoch {epoch+1:02d}/{epochs}  auc={score:.5f}  best={best_score:.5f}  "
                      f"ls={ls:.4f}  drop={drop:.4f}{mark}", flush=True)
            if p["use_early_stopping"]:
                patience = best_epoch * p["early_stopping_multiplicative_patience"] + p["early_stopping_additive_patience"]
                if (epoch + 1) > patience:
                    if verbose >= 1:
                        print(f"  Early stop at epoch {epoch+1} (best {best_epoch})", flush=True)
                    break
        if best_state is not None:
            self.model_.load_state_dict(best_state, strict=True)
        self.best_score_ = best_score
        self.best_val_probs_ = best_val_probs
        self._dev = dev

        if verbose >= 1:
            print(f" ===> best AUC: {best_score:.5f} (epoch {best_epoch})", flush=True)

        return self

    def predict_proba(self, X):
        eval_bs = self.params["eval_bs"]
        Xn = self.preproc_.transform(X[self.num_cols_].values.astype(np.float32))
        Xc = X[self.cat_cols_].values.astype(np.int64)
        Xc = np.clip(Xc, 0, np.array(self.cat_dims_) - 1)
        Xn_t = torch.as_tensor(Xn, dtype=torch.float32, device=self._dev)
        Xc_t = torch.as_tensor(Xc, dtype=torch.long, device=self._dev)
        self.model_.eval()
        with torch.no_grad():
            return np.concatenate([
                self.model_(Xn_t[s:s+eval_bs], Xc_t[s:s+eval_bs]).mean(dim=1).cpu().numpy()
                for s in range(0, len(Xn), eval_bs)
            ], axis=0)
    # Predict the class label using the highest predicted probability.
    def predict(self, X):
        return self.classes_[np.argmax(self.predict_proba(X), axis=1)]
```

# <p style="background:linear-gradient(120deg,#453548 0%,#60495D 24%,#806477 48%,#70837C 73%,#9A8B6B 100%);color:#FFFDF8;font-family:'Trebuchet MS','Verdana',sans-serif;font-size:24px;text-align:center;padding:22px 20px;border-radius:18px;width:78%;margin:34px auto;font-weight:800;letter-spacing:1.2px;line-height:1.3;text-shadow:0 2px 4px rgba(0,0,0,0.3);border:3px solid #E7D9C9;box-shadow:0 0 0 3px #493B4A,0 0 0 7px #C5A66A,0 10px 22px rgba(55,43,55,0.28),inset 0 2px 0 rgba(255,255,255,0.2);outline:3px solid #8E6B86;outline-offset:9px;">Let's Prepare Model Features</p>

```python
# Load training and test datasets
def load_data(train_path, test_path):
    train = pd.read_csv(train_path)
    test = pd.read_csv(test_path)
    # Define the identifier and target column.
    ID_COL = "id"
    TARGET = "addicted_label"
    X = train.drop([ID_COL, TARGET], axis=1).copy()
    y = train[TARGET]
    X_test = test.drop([ID_COL], axis=1).copy()
    test_ids = test[ID_COL]
    cat_cols = X.select_dtypes(include=["object"]).columns.tolist()
    num_cols = X.select_dtypes(exclude=["object"]).columns.tolist()
    print(f"Raw shapes  → train: {X.shape}  test: {X_test.shape}")
    print(f"Categorical → {len(cat_cols)}: {cat_cols}")
    print(f"Numerical   → {len(num_cols)}: {num_cols}")

    # Medians
    medians = {c: X[c].median() for c in num_cols}
    print("\nMissing rates & medians:")
    for c, m in medians.items():
        print(f"  {c:30s}  median={m:8.3f}  miss={X[c].isnull().mean()*100:5.1f}%")

    cat_map = {}
    # Build and fit engineered features
    X, new_cats, new_nums = build_features(
        X, list(cat_cols), list(num_cols), cat_map, fit=True, median_dict=medians
    )
    X_test, _, _ = build_features(
        X_test, list(cat_cols), list(num_cols), cat_map, fit=False, median_dict=medians
    )

    all_num = num_cols + new_nums
    all_cat = sorted(cat_cols + new_cats)

    # Keep column order consistent
    X = X.reindex(sorted(X.columns), axis=1)
    X_test = X_test.reindex(sorted(X_test.columns), axis=1)
    print(f"\nAfter FE ===> train: {X.shape}  test: {X_test.shape}")
    print(f"Final cat ===> {len(all_cat)}")
    print(f"Final num ===> {len(all_num)}  (new interactions: {len(new_nums)})")
    print(f"New numeric features: {new_nums}")
    return X, y, X_test, test_ids, all_cat, all_num
```

# <p style="background:linear-gradient(120deg,#453548 0%,#60495D 24%,#806477 48%,#70837C 73%,#9A8B6B 100%);color:#FFFDF8;font-family:'Trebuchet MS','Verdana',sans-serif;font-size:24px;text-align:center;padding:22px 20px;border-radius:18px;width:78%;margin:34px auto;font-weight:800;letter-spacing:1.2px;line-height:1.3;text-shadow:0 2px 4px rgba(0,0,0,0.3);border:3px solid #E7D9C9;box-shadow:0 0 0 3px #493B4A,0 0 0 7px #C5A66A,0 10px 22px rgba(55,43,55,0.28),inset 0 2px 0 rgba(255,255,255,0.2);outline:3px solid #8E6B86;outline-offset:9px;">Let's Execute the Training Pipeline & Generate the Submisison File</p>

```python
def main():
    t0 = time.time()
    TRAIN_PATH = "/kaggle/input/competitions/playground-series-s6e8/train.csv"
    TEST_PATH  = "/kaggle/input/competitions/playground-series-s6e8/test.csv"
    print("===========================================================")
    print("RealMLP: Smartphone Addiction (S6E8)")
    print(f"Multi-seed: {SEEDS}")
    print("===========================================================")
    X, y, X_test, test_ids, cat_cols, num_cols = load_data(TRAIN_PATH, TEST_PATH)
    print(f"\nDataset: {len(X)} train  |  {len(X_test)} test")
    oof_all = np.zeros((len(X), 2))
    test_all = np.zeros((len(X_test), 2))
    seed_scores = []
    for s_idx, seed in enumerate(SEEDS):
        seed_everything(seed)
        if torch.cuda.is_available():
            torch.cuda.manual_seed_all(seed)
        print(f"\n{'#'*25}")
        print(f"### SEED {s_idx+1}/{len(SEEDS)}  (seed={seed})")
        print(f"{'#'*25}")
        cfg = {**CFG, "random_state": seed}
        skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
        oof = np.zeros((len(X), 2))
        test_pred = np.zeros((len(X_test), 2))
        fold_aucs = []
        for fold, (tr_idx, va_idx) in enumerate(skf.split(X, y), 1):
            X_tr = X.iloc[tr_idx].copy()
            X_va = X.iloc[va_idx].copy()
            X_te = X_test.copy()
            y_tr = y.iloc[tr_idx]
            y_va = y.iloc[va_idx]

            # Target encoding 
            te_cols = [c for c in cat_cols if not c.endswith("_qbin") and not c.startswith("miss_")]
            if te_cols:
                te = TargetEncoder(cv=N_FOLDS, smooth="auto", shuffle=True, random_state=seed)
                X_tr[[f"te_{c}" for c in te_cols]] = te.fit_transform(X_tr[te_cols], y_tr)
                X_va[[f"te_{c}" for c in te_cols]] = te.transform(X_va[te_cols])
                X_te[[f"te_{c}" for c in te_cols]] = te.transform(X_te[te_cols])
            if fold == 1 and s_idx == 0:
                print(f"  Features after TE: {X_tr.shape[1]}")
                print(f"  Numeric: {len([c for c in X_tr.columns if c not in cat_cols])}")
                print(f"  Categorical: {len(cat_cols)}")

            print(f"\n  Fold {fold}/{N_FOLDS}", flush=True)
            t_fold = time.time()
            model = RealMLPClassifier(**cfg)
            model.fit(X_tr, y_tr, X_va, y_va, cat_col_names=cat_cols)
            oof[va_idx] = model.best_val_probs_
            test_pred += model.predict_proba(X_te) / N_FOLDS
            fold_auc = roc_auc_score(y_va, oof[va_idx, 1])
            fold_aucs.append(fold_auc)
            print(f"  Fold {fold} AUC: {fold_auc:.5f}  |  time: {(time.time()-t_fold)/60:.1f} min")
            del model
            torch.cuda.empty_cache()

        seed_auc = roc_auc_score(y, oof[:, 1])
        seed_scores.append(seed_auc)
        oof_all += oof
        test_all += test_pred
        print(f"\n  Seed {seed} OOF AUC: {seed_auc:.5f}")
        print(f"  Fold AUCs: {[f'{a:.5f}' for a in fold_aucs]}")

    # Average seeds
    oof_all /= len(SEEDS)
    test_all /= len(SEEDS)
    final_auc = roc_auc_score(y, oof_all[:, 1])
    print("===========================================================")
    print(f"Multi-seed OOF AUC : {final_auc:.5f}")
    print(f"Per-seed AUCs : {[f'{s:.5f}' for s in seed_scores]}")
    print(f"Avg of seeds : {np.mean(seed_scores):.5f}")
    print("===========================================================")

    # Submission
    print("\nGenerating submission...")
    sub = pd.DataFrame({
        "id": test_ids,
        "addicted_label": test_all[:, 1]
    })
    sub.to_csv("submission.csv", index=False)
    print(f"Submission shape: {sub.shape}")
    print(f"Pred stats → mean={sub['addicted_label'].mean():.4f}  "
          f"min={sub['addicted_label'].min():.4f}  max={sub['addicted_label'].max():.4f}")

    # Save OOF & test preds for later ensembling
    pd.DataFrame({"addicted_label": y, "oof_pred": oof_all[:, 1]}).to_csv("oof.csv", index=False)
    pd.DataFrame({"id": test_ids, "pred": test_all[:, 1]}).to_csv("test_pred.csv", index=False)

    elapsed = time.time() - t0
    print("===========================================================")
    print(f"DONE in {elapsed/60:.1f} minutes")
    print(f"OOF AUC: {final_auc:.5f}")
    print("===========================================================")

if __name__ == "__main__":
    main()
```
